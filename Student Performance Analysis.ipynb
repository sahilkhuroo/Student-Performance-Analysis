### 1. scatter plot
 - A scatter plot is a graph used to show the relationship between two numerical variables. Each dot represents one observation from the dataset. The X-axis shows one variable, and the Y-axis shows another variable. It helps us identify patterns and relationships between the variables. If the dots move upward, it shows a positive relationship, while downward movement shows a negative relationship. Randomly scattered dots may show little or no relationship. In Seaborn, we use sns.scatterplot() to create it. In simple words, a scatter plot uses dots to show the relationship between two values.

```python
sns . scatterplot(data=data, x="Hours_Studied",y= "Exam_Score")
plt.title("Hours studied vs Exam Score")
plt.show()
```

### 2. line plot
- A line plot is a graph used to show how data changes over time or in a specific order. It represents values using points and connects them with a line. The X-axis usually shows time, such as days, months, or years, while the Y-axis shows the measured value, such as marks, sales, or temperature. An upward line shows that the value is increasing, while a downward line shows a decrease. If the line moves up and down, it shows fluctuations. Line plots are useful for finding trends and patterns in data. In Seaborn, we use the sns.lineplot() function to create a line plot. In simple words, a line plot shows how data changes by connecting points with a line.


```python
sns.lineplot(data=data, x="Attendance", y="Exam_Score")
plt.title("Attendance vs Exam Score")
plt.show()
```

### 3. stripplot
 - A strip plot is a type of graph used to show individual data values for different categories. In a strip plot, each dot represents one observation or record from the dataset. The X-axis usually contains categories, while the Y-axis contains numerical values. For example, we can use a strip plot to compare the marks of students from different classes. It helps us understand how the data is distributed within each category. If many dots are close together, it means the values are similar. If the dots are spread out, it shows more variation in the data. We can use jitter=True to separate overlapping dots and make them easier to see. Strip plots are useful for comparing groups and identifying unusual values. In Seaborn, a strip plot is created using the sns.stripplot() function. In simple words, a strip plot shows individual data values as dots so we can easily understand the distribution of data.

```python
sns.stripplot(data=data,x="Parental_Involvement", y="Exam_Score",hue="Parental_Involvement",palette="Set2",legend=False)
plt.title("Parental Involvement vs Exam Score")
plt.show()
```

### 4. Heat Map
 - A heatmap is a graph that uses different shades of colour to represent values in a dataset. Darker or stronger colours usually represent higher values, while lighter colours represent lower values. It helps us quickly identify patterns, relationships, and differences in data. In Data Science, heatmaps are often used to show the correlation between numerical variables. Each cell represents the relationship between two variables. In Seaborn, we can create a heatmap using sns.heatmap(). We can use annot=True to display the actual values inside the cells. In simple words, a heatmap uses colours to make data patterns easy to understand.

```python
sns.heatmap(data.select_dtypes("number").corr(),annot=True,)
plt.title("Correlation Heatmap")
plt.xticks(rotation=45)
plt.yticks(rotation=50)
plt.show()
```
